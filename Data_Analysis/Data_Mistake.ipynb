```python
import pandas as pd
import numpy as np
```

```python
path  = "1m_ohlcv_5y_adjusted.csv"
df    = pd.read_csv(path)
```

#### 1.Missing values

```python
####################### Missing data #####################
df["Date-Time"] = pd.to_datetime(df["Date-Time"], errors="coerce")
missing_dates = df["Date-Time"].isna()
missing_open  = df["Open"].isna()
missing_high  = df["High"].isna()
missing_low   = df["Low"].isna()
missing_last  = df["Last"].isna()

missing_open_ad  = df["adj_open"].isna()
missing_high_ad  = df["adj_high"].isna()
missing_low_ad   = df["adj_low"].isna()
missing_last_ad  = df["adj_last"].isna()

missing_volume  = df["Volume"].isna()
missing_volume_ad = df["adj_volume"].isna()


print(missing_dates.sum())
duplicates = df["Date-Time"].duplicated()
print("Duplicate timestamps:", duplicates.sum())

print(missing_open.sum())
print(missing_high.sum())
print(missing_low.sum())
print(missing_last.sum())
print(missing_open_ad.sum())
print(missing_high_ad.sum())
print(missing_low_ad.sum())
print(missing_last_ad.sum())

print(missing_volume.sum())
print(missing_volume_ad.sum())
```

```python
all_missing_same_row = (missing_open & missing_high & missing_low & missing_last & missing_open_ad & missing_high_ad & missing_low_ad & missing_last_ad)
print("Rows where all 8 columns are missing:", all_missing_same_row.sum())
```

#### 2.Gaps in Time-Date

```python
MARKET_OPEN = "07:00"
MARKET_CLOSE = "12:30"
df = df.sort_values("Date-Time").reset_index(drop=True)

################ unique time
actual_times = pd.DatetimeIndex(df["Date-Time"].drop_duplicates())

################ Trading day
trading_days = sorted(df["Date-Time"].dt.date.unique())

############### expected time
expected_times = []

for day in trading_days:

    start = pd.Timestamp(f"{day} {MARKET_OPEN}", tz="UTC")
    end = pd.Timestamp(f"{day} {MARKET_CLOSE}", tz="UTC")

    day_minutes = pd.date_range(
        start=start,
        end=end,
        freq="1min"
    )

    expected_times.extend(day_minutes)

expected_times = pd.DatetimeIndex(expected_times)

############ Find missing bars
missing_times = expected_times.difference(actual_times)
missing_df = pd.DataFrame({"Missing_DateTime": missing_times})


print("=" * 50)
print("DATA AUDIT SUMMARY")
print("=" * 50)

print(f"Trading days found      : {len(trading_days):,}")
print(f"Expected minute bars    : {len(expected_times):,}")
print(f"Actual minute bars      : {len(actual_times):,}")
print(f"Missing minute bars     : {len(missing_times):,}")

########## file
missing_df.to_csv("missing_bars.csv", index=False)

print("\nSaved missing bars to:")
print("missing_bars.csv")

print("\nFirst 20 missing bars:")
print(missing_df.head(20))
```
