```python
import pandas as pd
import numpy as np
```

```python
path  = "1m_ohlcv_5y_adjusted.csv"
df    = pd.read_csv(path)
```

```python
####################### Missing data #####################
df["Date-Time"] = pd.to_datetime(df["Date-Time"], errors="coerce")
missing_dates = df["Date-Time"].isna()
missing_open  = df["Open"].isna()
missing_high  = df["High"].isna()
missing_low   = df["Low"].isna()
missing_last  = df["Last"].isna()

missing_open_ad  = df["adj_open"].isna()
missing_high_ad  = df["adj_high"].isna()
missing_low_ad   = df["adj_low"].isna()
missing_last_ad  = df["adj_last"].isna()

missing_volume  = df["Volume"].isna()
missing_volume_ad = df["adj_volume"].isna()


print(missing_dates.sum())
duplicates = df["Date-Time"].duplicated()
print("Duplicate timestamps:", duplicates.sum())

print(missing_open.sum())
print(missing_high.sum())
print(missing_low.sum())
print(missing_last.sum())
print(missing_open_ad.sum())
print(missing_high_ad.sum())
print(missing_low_ad.sum())
print(missing_last_ad.sum())

print(missing_volume.sum())
print(missing_volume_ad.sum())
```

```python
all_missing_same_row = (missing_open & missing_high & missing_low & missing_last & missing_open_ad & missing_high_ad & missing_low_ad & missing_last_ad)
print("Rows where all 8 columns are missing:", all_missing_same_row.sum())
```
